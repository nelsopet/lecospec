```r
source("Functions/lecospectR.R", echo = FALSE, verbose = FALSE)
library(torch)
library(luz)
```

```r
model_files <- c(
    "dl_jigsaw_v4.dat",
    "dl_linear_v1_512.dat",
    "dl_linear_v2_512.dat",
    "dl_resnet_v0.dat"
)
```

```r
for(f in model_files){
    model_id <- uuid::UUIDgenerate()
    model <- luz::luz_load(f)
    model_dir <- paste0("./temp/", model_id, "/")
    print(model)
    results <- validate_model(
        ml_model = model,
        bandwidth = 1,
        save_directory = model_dir
        )

    aggregated_results <- aggregate_results(directory = model_dir)
    add_model_to_manifest(
        model_id = model_id,
        hyperparam1 = f, 
        accuracy = "See Attached",
        r2 = calculate_r_squared(df = aggregated_results))

}
```
